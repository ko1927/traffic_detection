```python
import cv2
import numpy as np

# 이미지 경로
image_path = './3.png'  # ← 파일명 수정 (예: '1.png')
image = cv2.imread(image_path)

# HSV 및 Gray 변환
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# 밝기 threshold 적용 (밝은 영역)
_, mask_bright = cv2.threshold(gray, 180, 255, cv2.THRESH_BINARY)

# --- 색상별 HSV 범위 설정 ---

# 🔵 파란불 (청록 포함)
lower_blue = np.array([80, 100, 180])
upper_blue = np.array([100, 255, 255])
mask_blue = cv2.inRange(hsv, lower_blue, upper_blue)
mask_blue = cv2.bitwise_and(mask_blue, mask_bright)
result_blue = cv2.bitwise_and(image, image, mask=mask_blue)

# 🟡 노란불
lower_yellow = np.array([20, 100, 180])
upper_yellow = np.array([35, 255, 255])
mask_yellow = cv2.inRange(hsv, lower_yellow, upper_yellow)
mask_yellow = cv2.bitwise_and(mask_yellow, mask_bright)
result_yellow = cv2.bitwise_and(image, image, mask=mask_yellow)

# 🔴 빨간불 (Hue는 양 끝에 걸쳐 있어서 두 범위 필요)
lower_red1 = np.array([0, 100, 180])
upper_red1 = np.array([10, 255, 255])
lower_red2 = np.array([160, 100, 180])
upper_red2 = np.array([180, 255, 255])
mask_red1 = cv2.inRange(hsv, lower_red1, upper_red1)
mask_red2 = cv2.inRange(hsv, lower_red2, upper_red2)
mask_red = cv2.bitwise_or(mask_red1, mask_red2)
mask_red = cv2.bitwise_and(mask_red, mask_bright)
result_red = cv2.bitwise_and(image, image, mask=mask_red)

# --- 결과 출력 ---
cv2.imshow("Original", image)
cv2.imshow("Red Light", result_red)
cv2.imshow("Yellow Light", result_yellow)
cv2.imshow("Blue Light", result_blue)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
